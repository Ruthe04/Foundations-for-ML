# **Deductive Reasoning**
***
Deductive reasoning is a form of logical reasoning that we often use in everyday situations without perceiving it. For example: <br><br> Today, I will study either biology or physics <br> I will not study biology today <br> Therefore, I will study physics today. <br><br> This form of reasoning can be represented as: <br> $P$ or $Q$ <br> Not $P$ <br> Therefore, $Q$ <br> Using this pattern, we can recognize whether an argument is valid by analyzing its statements (premises and conclusion) and checking whether the conclusion logically follows from the premises. 

### Logical Symbols
In formal logic, each connective is represented by a specific symbol, and the conclusions are indicated using inference symbols: <center> <br> $\neg$  →  not <br> $\land$  →  and <br> $\lor$  →  or <br> $\therefore$ → conclusion marker </center> <br> With the formal symbols, we can rewrite the argument from our first example: <center> <br> $P$ $\lor$ $Q$ <br> $\neg$ $P$ <br> $\therefore$ $Q$ </center> <br> Where $P$ represents "I will study biology today" and $Q$ represents "I will study physics today"  <br> 

***Note:** The connectives $\land$ (and $\lor$) can only be used between statements; they cannot be applied to every use of the word "and" or "or", for example: "My friends and I are going". Likewise, the symbol $\neg$ can only be placed before a statement to negate it. <br> <br>

**Examples with symbols** <br> $A$ $\lor$ $B$ → Where A stands for "Andrew is my husband" and B for "Bill is my husband", the statement can be understood as "Either Andrew or Bill is my husband" (or both). <br> 

***Note:** The connective $\lor$ (or), in everyday language, can mean that just one statement is true. In formal logic, however, it usually represents the inclusive "or", meaning that one or both statements may be true.

***Note:** Sometimes, logical words are hidden within mathematical notation; for example, $2 \leq \pi$. When read aloud, the connective $or$ appears. Therefore, we can rewrite that statement as $(2<\pi) \lor (2 = \pi)$

#### **Exercises**  
$1$. Rewrite the logical statements using the logical symbols: <br> 
* **Statement a**: We’ll have either a reading assignment or homework problems, but we won’t have both homework problems and a test. <br>
  Answer: Let R represent "reading assignment", H represent "homework problems", and T represent "a test". <br> 
   $(R \lor H) \land \neg(H \land T)$ <br>

  
    ***Note:** The word $but$ usually represents an addition of information; therefore, it can be rewritten using $\land$. <br>

* **Statement b**: You won’t go skiing, or you will and there won’t be any snow. <br>
  Answer: Let S represent "you will go skiing" and C represent "there won’t be any snow".<br>
  $\neg S \lor (S \land C)$ <br>

* **Statement c:** Either John and Bill are both telling the truth, or neither of them is. <br>
  Answer: Let J represent "John is telling the truth", and B represent "Bill is telling the truth". <br>
  $(J \land B) \lor (\neg J \land \neg B)$ <br>

  ***Note:** When the negation applies to each statement, we need to add the negation symbol to each one. <br>
  
  
$2$. Let S stand for the statement “Steve is happy” and G for “George is happy.” What English sentences are represented by the following expressions? <br>  <center>
(a) (S ∨ G) ∧ (¬S ∨ ¬G) <br>
Answer: Either Steve or George is happy, but not both. <br>
(b) [S ∨ (G ∧ ¬S)] ∨ ¬ G <br>
Answer: Steve is happy, or (George is happy and Steve is not happy), or George is not happy. <br> 
(c) S ∨ [G ∧ (¬S ∨ ¬G)] <br>
Answer: Steve is happy, or (George is happy and (Steve is not happy or George is not happy)). <br>
</center>

## **Table of symbols and their meaning**  

| Symbol | Meaning | Example |
|:---:|:---:|:---:|

| $\subsetneq$ | Proper subset (A is in B, but B has extra) | $A \subsetneq B$ |
| $\varnothing$ | Empty set | $A = \varnothing$ |
| $\sum$ | Summation | $\sum_{i=1}^{4} i$ |
